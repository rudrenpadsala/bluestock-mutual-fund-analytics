```python
# =====================================
# STEP 13.1: Prepare NAV Data
# =====================================

# Convert date column
nav_history["date"] = pd.to_datetime(nav_history["date"])

# Sort data
nav_history = nav_history.sort_values(["amfi_code", "date"])

print(nav_history.head())
```

```python
# =====================================
# STEP 13.2: Create Pivot Table
# =====================================

pivot_nav = nav_history.pivot_table(
    index="date",
    columns="amfi_code",
    values="nav"
)

print(pivot_nav.head())
```

```python
# =====================================
# STEP 13.3: Select 10 Funds
# =====================================

# Select the first 10 AMFI codes
selected_funds = pivot_nav.columns[:10]

# Create a new DataFrame with only these 10 funds
pivot_nav = pivot_nav[selected_funds]

print("Selected AMFI Codes:")
print(selected_funds)

print("\nPreview:")
display(pivot_nav.head())
```

```python
# =====================================
# STEP 13.4: Calculate Daily Returns
# =====================================

daily_returns = pivot_nav.pct_change()

# Remove the first row containing NaN values
daily_returns = daily_returns.dropna()

print(daily_returns.head())
```

```python
# =====================================
# STEP 13.5: Correlation Matrix
# =====================================

correlation_matrix = daily_returns.corr()

print(correlation_matrix)
```

```python
# =====================================
# STEP 13.6: Correlation Heatmap
# =====================================

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12,8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    linewidths=0.5,
    fmt=".2f"
)

plt.title(
    "Correlation Matrix of Daily NAV Returns",
    fontsize=16
)

plt.tight_layout()

plt.show()
```

```python
# =====================================
# STEP 13.7: Save Heatmap
# =====================================

plt.figure(figsize=(12,8))

sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    annot=True,
    fmt=".2f",
    linewidths=0.5
)

plt.title("NAV Return Correlation Matrix")

plt.tight_layout()

plt.savefig(
    "../reports/nav_correlation_heatmap.png",
    dpi=300
)

plt.close()

print("Correlation Heatmap Saved Successfully!")
```

```python
# =====================================
# STEP 14.2: Aggregate Sector Weights
# =====================================

sector_weights = (
    portfolio
    .groupby("sector")["weight_pct"]
    .sum()
    .sort_values(ascending=False)
)

print(sector_weights)
```

```python
# =====================================
# STEP 14.3: Sector Allocation Donut Chart
# =====================================

import matplotlib.pyplot as plt

plt.figure(figsize=(10,10))

plt.pie(
    sector_weights,
    labels=sector_weights.index,
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops={"width":0.4}
)

plt.title(
    "Sector Allocation Across Equity Funds",
    fontsize=16
)

plt.tight_layout()

plt.show()
```

```python
# =====================================
# STEP 14.4: Top 10 Sectors
# =====================================

top10 = sector_weights.head(10)

plt.figure(figsize=(12,6))

plt.bar(
    top10.index,
    top10.values
)

plt.xticks(rotation=45)

plt.title(
    "Top 10 Sectors by Portfolio Weight",
    fontsize=16
)

plt.xlabel("Sector")

plt.ylabel("Weight (%)")

plt.tight_layout()

plt.show()
```
