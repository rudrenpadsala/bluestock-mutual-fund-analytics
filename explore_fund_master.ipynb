```python
import pandas as pd

# Load Fund Master Dataset
fund_master = pd.read_csv("data/raw/01_fund_master.csv")

print("=" * 80)
print("FUND MASTER DATASET ANALYSIS")
print("=" * 80)

# Basic Information
print("\n[1] Dataset Shape")
print(fund_master.shape)

print("\n[2] Column Names")
print(fund_master.columns.tolist())

print("\n[3] Data Types")
print(fund_master.dtypes)

print("\n[4] First 5 Rows")
print(fund_master.head())

print("\n[5] Missing Values")
print(fund_master.isnull().sum())

print("\n[6] Duplicate Rows")
print(fund_master.duplicated().sum())

# Explore All Columns
print("\n" + "=" * 80)
print("UNIQUE VALUES IN EACH COLUMN")
print("=" * 80)

for col in fund_master.columns:

    try:
        unique_count = fund_master[col].nunique()

        print(f"\nColumn: {col}")
        print(f"Unique Values Count: {unique_count}")

        if unique_count <= 20:
            print(fund_master[col].unique())

    except:
        pass

# Fund House Analysis
possible_fund_house_cols = [
    "fund_house",
    "amc_name",
    "fundhouse",
    "asset_management_company"
]

for col in possible_fund_house_cols:

    if col in fund_master.columns:

        print("\n" + "=" * 80)
        print("UNIQUE FUND HOUSES")
        print("=" * 80)

        print(fund_master[col].unique())
        print("\nTotal Fund Houses:", fund_master[col].nunique())

# Category Analysis
possible_category_cols = [
    "category",
    "scheme_category",
    "fund_category"
]

for col in possible_category_cols:

    if col in fund_master.columns:

        print("\n" + "=" * 80)
        print("UNIQUE CATEGORIES")
        print("=" * 80)

        print(fund_master[col].unique())
        print("\nTotal Categories:", fund_master[col].nunique())

# Sub Category Analysis
possible_subcategory_cols = [
    "subcategory",
    "sub_category",
    "scheme_sub_category"
]

for col in possible_subcategory_cols:

    if col in fund_master.columns:

        print("\n" + "=" * 80)
        print("UNIQUE SUB-CATEGORIES")
        print("=" * 80)

        print(fund_master[col].unique())
        print("\nTotal Sub-Categories:", fund_master[col].nunique())

# Risk Grade Analysis
possible_risk_cols = [
    "risk_grade",
    "riskometer",
    "risk",
    "risk_level"
]

for col in possible_risk_cols:

    if col in fund_master.columns:

        print("\n" + "=" * 80)
        print("UNIQUE RISK GRADES")
        print("=" * 80)

        print(fund_master[col].unique())
        print("\nTotal Risk Grades:", fund_master[col].nunique())

print("\n" + "=" * 80)
print("ANALYSIS COMPLETED")
print("=" * 80)
```
